```python
import sympy as sp

# 定義變數
x = sp.symbols('x')

# 定義函數
f = sp.sin(x)**2 * sp.cos(x)**3

# 微分
f_prime = sp.diff(f, x)

# 積分
f_integral = sp.integrate(f_prime, x)

print("原函數 f(x) =")
sp.pprint(f)

print("\n導數 f'(x) =")
sp.pprint(f_prime)

print("\n積分 ∫f'(x)dx =")
sp.pprint(f_integral)
```

```python


# 定義變數
x = sp.symbols('x')

# 定義原函數 f(x)
f = sp.sin(x)**2 * sp.cos(x)**3

# 先求導數 f'(x)
f_prime = sp.diff(f, x)

# 再對 f'(x) 積分
result = sp.integrate(f_prime, x)

# 輸出結果
print("f'(x) =")
sp.pprint(f_prime)

print("\n∫ f'(x) dx =")
sp.pprint(result)
```
